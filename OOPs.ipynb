# Class and objects
* **Class:** The blueprint of the real world entities
* **Objects:** These are the instance of the class that access the properties(Attribute and methods) of class

* Whenever you are going to create the class, you will access that class with the help objects.
* Objects we always create outside of the class
* Inside of the class we will create Attributes(properties) and methods(functionalities)


```python
# To Create a class | Keyword : class (LowerCase)
class Student:          # Always create the class name starting Capital letter
  name= "Tonney"         # Attributes
  age= 27              # Attributes

  def display(self):       # Keyword alert : self -> Connector between my object to data member / Attributes and methods
    print(self.name)    # to access the attribute inside the class we use self.attrName
    print(self.age)
```

```python
# To create an object
ob1 = Student() # Call your construc
print(ob1.name)
print(ob1.age)
ob1.display()
ob2 = Student()
print(ob2.name)
ob2.name = "Aryan"
print(ob2.name)
```

```python
# Constructor -> It initialize the data member that we have in the class when we create the object
# Whenever we create the method inside the class, The first parameter will always be self
# To create the constructor : keyword alert __init__() -> Magic methods
```

```python
class Database:
  def __init__(self,empName,empDomain,empSalary):
    self.empName = empName      # self.empName - Attribute
    self.empDomain = empDomain
    self.empSalary = empSalary

  def display(self):
    print(f"Name : {self.empName} , Domain : {self.empDomain}, salary : {self.empSalary}")
```

# Single Inheritance
* **one parent class -------- one child class**

```python
# Single inheritance
class Parent:
  def __init__(self):
    self.name = "Aryan"  # this attribute should be accessed by my child

class Child(Parent):      # Child class inherited from parent class
  def __init__(self):
    Parent.__init__(self) # ParentClassName.__init__(self,para) -> To call the parent class constructor in your child class
  def display(self):
    print(self.name)
```

```python
c1 = Child()
c1.display()
```

# Multilevel inheritance
* **parent class ------ Child class ------- Child2**


```python
class Parent:               # Parent class
  def __init__(self,name):
    self.name = name

class Child(Parent):      # Child class inherited from parent class
  def __init__(self,name,age):
    Parent.__init__(self,name)
    self.age = age            # This attribute is of my Child class

  def hello(self):
    print('Inherited from the Child class')

class Child2(Child):       # Child2 class inherits from Child class
  def __init__(self,name,age):
    Child.__init__(self,name,age)

  def display(self):
    print(self.name, self.age)
```

```python
ob1 = Child2("Aryan",28)
ob1.display()
ob1.hello()
ob = Parent("XYZ")
ob.name
```

# Multiple Inheritance
* **Parent 1 + Parent 2 --------- Child**

```python
class Parent1:
  def __init__(self):
    self.name = "Shashank"

class Parent2:
  def __init__(self):
    self.age = 30

class Child(Parent1,Parent2):
  def __init__(self):
    self.salary = 200000
    Parent1.__init__(self)
    Parent2.__init__(self)

  def Display(self):
    print(self.name,self.age,self.salary)

ob1 = Child()
ob1.Display()
```

# Hierarchical Inheritane
* **Parent Class -------- Child 1 + Child2**


```python
class Parent:
  def __init__(self):
    self.name = "Aryan"

class Child1(Parent):
  def __init__(self):
    Parent.__init__(self)
    self.age = 30
  def Display(self):
    print(self.name,self.age)

class Child2(Parent):
  def __init__(self):
    Parent.__init__(self)
    self.salary = 300000

  def Display(self):
    print(self.name,self.salary)
```

```python
ob1 = Child1()
ob1.Display()
ob2 = Child2()
ob2.Display()
```

```python
# Create the class Maths
# contructor ( num1 and num2)
# inherit this class Math to class Sum
# create one method under class Sum (num1 + num2)

# self we will use whener we create attribute and first parameter of method
```

```python
class Maths:
  def __init__(self,num1,num2):
    self.num1 = num1
    self.num2 = num2

class Sum(Maths):
  def __init__(self,num1,num2):
    Maths.__init__(self,num1,num2)

  def Display(self):
    print(self.num1 + self.num2)
```

```python
ob1 = Sum(2,3)
ob1.Display()
```

# Questions:
* What is method overloading : Same name ---- diff funct
* Why we cannot perform method overloading : Python is an interpreter language
* What is the role of super() : Access the function from super class

```python
# Super() : Calls the function from Parent class
class Parent:
  def __init__(self):
    self.name = "Square"
    self.sides = 4

  def area(self):
    print(f"the area of square is :{self.sides**2}")

class Child(Parent):
  def __init__(self):
    Parent.__init__(self)
    self.name = "Rectangle" # Overriding
    self.width = 10
    self.length = 5

  def area(self): # My area method from parent class was over written
    # super().area()
    Parent.area(self)
    print(f"The area of rectangle is: {self.width * self.length}")

  def parentArea(self):
    super().area() # Calls the parent class Area method
# With this super() We have called the area method from parent class by creating another function (ParentArea)
```

```python
ob1 = Child()
ob1.area()
# ob1.parentArea()
```

```python
ob = Parent()
ob.area()
```
